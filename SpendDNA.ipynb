# **Feature 8**

````python
total_debit_amount = debit_df['Amount'].sum()

def pct_of_debits(mask):
    return debit_df.loc[mask, 'Amount'].sum() / total_debit_amount * 100

print(f"Total debit amount: ₹{total_debit_amount:,.0f}")
````

````python
def is_foodie():
    mask = debit_df['category'].isin(['Food Delivery', 'Restaurants', 'Cafe'])
    pct = pct_of_debits(mask)
    return pct > 25, pct

def is_quick_commerce_junkie():
    pct = pct_of_debits(debit_df['category'] == 'Quick Commerce')
    return pct > 15, pct

def is_shopaholic():
    pct = pct_of_debits(debit_df['category'] == 'E-commerce')
    return pct > 15, pct

def is_investor():
    pct = pct_of_debits(debit_df['category'] == 'Investments')
    return pct > 15, pct

def is_late_night_snacker():
    food = debit_df[debit_df['category'] == 'Food Delivery']
    late = food[(food['hour'] >= 21) | (food['hour'] <= 2)]
    pct = len(late) / len(food) * 100 if len(food) > 0 else 0
    return pct > 50, pct

def is_cab_commuter():
    pct = pct_of_debits(debit_df['category'] == 'Transport')
    return pct > 10, pct

def is_subscription_lover():
    sub_vendors = debit_df.loc[debit_df['category'] == 'Subscriptions', 'vendor_clean'].nunique()
    return sub_vendors >= 5, sub_vendors

def is_yolo_spender():
    savings_rate = (total_credits - total_debits) / total_credits * 100
    return savings_rate < 10, savings_rate

def is_disciplined_saver():
    savings_rate = (total_credits - total_debits) / total_credits * 100
    return savings_rate > 40, savings_rate
````

````python
archetypes = {
    'The Foodie': is_foodie(),
    'The Quick Commerce Junkie': is_quick_commerce_junkie(),
    'The Shopaholic': is_shopaholic(),
    'The Investor': is_investor(),
    'The Late-Night Snacker': is_late_night_snacker(),
    'The Cab Commuter': is_cab_commuter(),
    'The Subscription Lover': is_subscription_lover(),
    'The YOLO Spender': is_yolo_spender(),
    'The Disciplined Saver': is_disciplined_saver(),
}

matched = []
for name, (qualifies, metric) in archetypes.items():
    tag = "MATCH" if qualifies else "no match"
    print(f"[{tag:8}] {name:28} -> metric = {metric:.1f}")
    if qualifies:
        matched.append(name)

print(f"\nRahul matches {len(matched)} archetype(s): {', '.join(matched)}")
````

````python
# bonus rule archetype:
def is_pavement_coffee_connoisseur():
    """
    THE PAVEMENT COFFEE CONNOISSEUR:
    5+ Cafe transactions before 9 AM (grabbing coffee/chai en route to work —
    classic Bengaluru tech commute pattern) AND average ticket size under Rs.250
    (street-side chai stall, not a premium cafe chain sit-down order).
    """
    cafe = debit_df[debit_df['category'] == 'Cafe']
    early_cafe = cafe[cafe['hour'] < 9]
    avg_ticket = early_cafe['Amount'].mean() if len(early_cafe) > 0 else 0
    qualifies = len(early_cafe) >= 5 and avg_ticket < 250
    return qualifies, (len(early_cafe), avg_ticket)


custom_qualifies, custom_metric = is_pavement_coffee_connoisseur()
tag = "MATCH" if custom_qualifies else "no match"
print(f"[{tag}] The Pavement Coffee Connoisseur -> early-cafe count={custom_metric[0]}, avg ticket=Rs.{custom_metric[1]:.0f}")
````

# **Bonus Features**

````python
# Invented archetype:
def is_bill_splitter_bro():
    p2p = debit_df[debit_df['category'] == 'Personal Transfer'].copy()
    p2p['contact'] = p2p['Description'].str.extract(r'UPI-([A-Za-z]+)-')
    distinct_contacts = p2p['contact'].nunique()
    median_amount = p2p['Amount'].median() if len(p2p) > 0 else 0
    qualifies = distinct_contacts >= 5 and median_amount < 3000
    return qualifies, (distinct_contacts, median_amount)

custom_qualifies, custom_metric = is_bill_splitter_bro()
tag = "MATCH" if custom_qualifies else "no match"
print(f"[{tag}] The Bill-Splitter Bro -> {custom_metric[0]} distinct contacts, median transfer = Rs.{custom_metric[1]:.0f}")
````
